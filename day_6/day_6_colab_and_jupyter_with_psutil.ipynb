## Практика

Создам список отзывов и отфильтрую только хорошие.

```python
reviews = [
    {"text": "Отличный товар!", "rating": 5},
    {"text": "Так себе.", "rating": 2},
    {"text": "Нормально.", "rating": 3},
    {"text": "Очень понравилось.", "rating": 4},
]

positive_reviews = [
    review for review in reviews if review["rating"] >= 4
]

positive_reviews
```

В ячейке выше мы отфильтровали отзывы, у которых рейтинг больше или равен 4.

```python
import psutil

print(f"CPU count: {psutil.cpu_count()}")
print(f"Total memory: {psutil.virtual_memory().total / 1024**3:.2f} GB")
print(f"Available memory: {psutil.virtual_memory().available / 1024**3:.2f} GB")
```
